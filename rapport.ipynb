# À quelques kilomètres près : lycées et géographie des cadres dans trois aires urbaines de province

```python
# Le SSPCloud définit MPLBACKEND=Agg : sans cette ligne, les figures ne s'affichent pas
%matplotlib inline
from IPython.display import display

from cadres_lycees import accessibilite, analyse, collecte, config, figures, geo, nettoyage
```

## 3. Trois aires, trois profils

### 3.1 Pourquoi ces trois aires

```python
referentiel = geo.referentiel_communes(config.AIRES_CANDIDATES)
annuaire = collecte.annuaire_education()
lycees = nettoyage.lycees_geo(annuaire, collecte.ips_lycees(), referentiel)
profil = analyse.profil_aires(lycees)
```

```python
figures.fig1(profil)
```

### 3.3 La géographie de l'accès aux lycées

```python
acces_prive = accessibilite.indicateurs(
    referentiel, lycees[lycees["secteur"] == "prive"], "lyc_prive"
)
communes = referentiel.merge(
    acces_prive, left_on="code_commune", right_index=True, how="left", validate="one_to_one"
)
```

```python
for aire in config.AIRES_CANDIDATES:
    display(figures.fig3(communes[communes["aire"] == aire], lycees))
```
